# Imports & Project Setup

```python
from pathlib import Path
import pandas as pd
import numpy as np

PROJECT_ROOT = Path.cwd().resolve().parent if Path.cwd().resolve().name == 'notebooks' else Path.cwd().resolve()
```

# Load Dataset

```python
DATA_PATH = PROJECT_ROOT / 'data/processed/cleaned_dataset.csv'
TABLEAU_READY_PATH = PROJECT_ROOT / 'data/processed/tableau_ready_dataset.csv'

df = pd.read_csv(DATA_PATH)

print("Shape:", df.shape)
df.head()
```

# Basic KPI Calculations

```python
total_listings = df['id'].count()
avg_price = df['price'].mean()
median_price = df['price'].median()
avg_mileage = df['odometer'].mean()

print("Total Listings:", total_listings)
print("Average Price:", round(avg_price, 2))
print("Median Price:", round(median_price, 2))
print("Average Mileage:", round(avg_mileage, 2))
```

# Feature Engineering

```python
current_year = 2021
df['car_age'] = current_year - df['year']
```

```python
def price_category(price):
    if price < 8000:
        return "Budget"
    elif price < 20000:
        return "Mid-range"
    else:
        return "Premium"

df['price_category'] = df['price'].apply(price_category)
```

```python
df['type_upper'] = df['type'].str.upper()
df['fuel_upper'] = df['fuel'].str.upper()
df['condition_upper'] = df['condition'].str.upper()
```

```python
top_manufacturer = df['manufacturer'].value_counts().idxmax()
print("Top Manufacturer:", top_manufacturer)
```

```python
leading_state = df['state'].value_counts().idxmax()
print("Leading State:", leading_state)
```

```python
dominant_type = df['type'].value_counts().idxmax()
print("Dominant Type:", dominant_type)
```

```python
top_fuel = df['fuel'].value_counts().idxmax()
print("Top Fuel:", top_fuel)
```

# Aggregated Tables

```python
price_by_condition = df.groupby('condition')['price'].mean().reset_index()
price_by_condition.head()
```

```python
state_distribution = df['state'].value_counts().reset_index()
state_distribution.columns = ['state', 'count']
state_distribution.head()
```

```python
price_by_cylinders = df.groupby('cylinders')['price'].mean().reset_index()
price_by_cylinders.head()
```

```python
TABLEAU_READY_PATH.parent.mkdir(parents=True, exist_ok=True)
df.to_csv(TABLEAU_READY_PATH, index=False)

print(f"Saved Tableau-ready dataset to {TABLEAU_READY_PATH}")
```
